#Data Cleaning


```python
print(df.info())
```

### Transforming column "Total Charges" Data Type from object to float64

```python
df['Total Charges'] = pd.to_numeric(df['Total Charges'], errors = 'coerce')
```

```python
print(df['Total Charges'].dtype)
```

```python
missing_total = df['Total Charges'].isnull().sum()
```

```python
print(f"Total new customers with no charges = {missing_total}")
```

###Fill in the empty boxes in Total Charges with the corresponding value in Monthly Charges

```python
df['Total Charges']=df['Total Charges'].fillna(df['Monthly Charges'])
```

```python
print(df['Total Charges'].isnull().sum())
```

#EDA

###General Churn percentage in the company

```python
print(round(df['Churn Label'].value_counts(normalize=True) * 100,2))
```

###Top 5 Reasons Why Customers Leave (Churn Reasons)

```python
top_reasons = df["Churn Reason"].value_counts().head(5)
```

```python
print(top_reasons)
```

###The relationship between contract type and customer departure (Churn)

```python
contract_churn = round(pd.crosstab(df['Contract'],df['Churn Label'],normalize='index')*100,2)
```

```python
print(contract_churn)
```

###Comparison of average monthly expenses between the overall average,customers who stayed with the company and those who left the company

```python
Avg_Mounthly_Charges = round(df['Monthly Charges'].mean(),2)
grouped_means = df.groupby('Churn Label')['Monthly Charges'].mean()
print(f"Avg Monthly Charges = ${Avg_Mounthly_Charges:.2f}")
print(f"Avg customer stayed = ${grouped_means['No']:.2f}")
print(f"Avg customer churn = ${grouped_means['Yes']:.2f}")
```

###The relationship between internet service and customer departure (Churn)

```python
internet_churn = round(pd.crosstab(df['Internet Service'],df['Churn Label'],normalize='index')*100,2)
```

```python
print(internet_churn)
```
